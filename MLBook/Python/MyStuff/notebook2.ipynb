```python
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.preprocessing import PolynomialFeatures # optimizations
from sklearn.ensemble import HistGradientBoostingRegressor, RandomForestRegressor
import joblib

housing = datasets.fetch_california_housing()

X = housing.data
y = housing.target

poly = PolynomialFeatures()
X = poly.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=432 # like a rng seed
)

model = HistGradientBoostingRegressor(
    max_iter= 200,
    learning_rate= 0.1,
)

model.fit(X_train, y_train)
# joblib.dump(model, "house_model.joblib")

prediction = model.predict(X_test)
score = r2_score(y_test, prediction)
print(score)

# 200 max, 0.1 learning

#outputs when testing models, used a loop through of models to find best one
#0.5613546734924131 LinearRegression()
#0.8387083453971109 HistGradientBoostingRegressor()
#0.80470186530567 RandomForestRegressor(n_jobs=-1)
```
