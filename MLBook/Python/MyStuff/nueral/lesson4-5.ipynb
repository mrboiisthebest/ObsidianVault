Get a image of a window!

````python
import pywinctl as pwc
import mss
import numpy as np
import cv2

TARGET_WINDOW = "Friends - Discord"
win = pwc.getWindowsWithTitle(TARGET_WINDOW)[0]
sct = mss.MSS()

while True:
    window_data = {
    "left": win.left,
    "top": win.top,
    "width": win.width,
    "height": win.height,
    }

    img = sct.grab(window_data)
    frame = np.array(img)
    frame = cv2.cvtColor(frame, cv2.COLOR_BGRA2BGR)

    cv2.imshow("image", frame)
    
    if cv2.waitKey(0):
        break
cv2.destroyAllWindows()
````

Resize and grayscale the images to pass into the ai

````python
import pywinctl as pwc
import mss
import numpy as np
import cv2

TARGET_WINDOW = "Friends - Discord"
win = pwc.getWindowsWithTitle(TARGET_WINDOW)[0]
sct = mss.MSS()

while True:
    window_data = {
    "left": win.left,
    "top": win.top,
    "width": win.width,
    "height": win.height,
    }

    img = sct.grab(window_data)
    
    frame = np.array(img)
    frame = cv2.cvtColor(frame, cv2.COLOR_BGRA2GRAY) # Turn into black and white
    frame = cv2.resize(frame, (128, 128), interpolation=cv2.INTER_AREA) # makes image smaller

    cv2.imshow("image", frame)
    
    if cv2.waitKey(0):
        break
cv2.destroyAllWindows()
````
