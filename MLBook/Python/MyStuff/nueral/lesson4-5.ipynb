**pywinctl**
used to get window and details about it

using WSL: sudo apt install -y x11-xserver-utils x11-utils

````python
import pywinctl as pwc

try:
    windows = pwc.getAllWindows()
    for w in windows:
        title = (w.title or "").strip()
        if not title:
            continue
        print(f"{title} -> x={w.left}, y={w.top}, w={w.width}, h={w.height}")
except SystemExit:
    print("PyWinCtl could not access the display server. Check X11 tools/xrandr/xset in WSL.")
````
