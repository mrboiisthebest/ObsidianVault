**pywinctl**
used to get window and details about it

using WSL: sudo apt install -y x11-xserver-utils x11-utils

Constanly shows a window in this case its the notepad!

````python
import pywinctl as pwc
import mss
import numpy as np
import mss.tools
import cv2

window = pwc.getWindowsWithTitle("Untitled - Notepad")[0]


with mss.MSS() as sct:
    while True:
        monitor_region = {
        "top": window.top,
        "left": window.left,
        "width": window.width,
        "height": window.height,
        }
        monitor = sct.monitors[0]

        screenshot = sct.grab(monitor_region)
        
        frame = np.array(screenshot)
        frame = cv2.cvtColor(frame, cv2.COLOR_BGRA2BGR)
        cv2.imshow("Dynamic AI Vision Stream", frame)

        if cv2.waitKey(1) & 0xFF == ord('q'):
            break
    cv2.destroyAllWindows()
````
