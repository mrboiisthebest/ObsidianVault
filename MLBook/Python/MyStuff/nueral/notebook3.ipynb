````python
X_train, X_test = X_train / 255.0, X_test / 255 # pre processing making numbers 0 - 1
````

````python
from tensorflow.keras import models, layers, losses, optimizers
````

````python
model = models.Sequential()
model.add(layers.InputLayer((28, 28)))
model.add(layers.Flatten())
model.add(layers.Dense(128, activation='relu'))
model.add(layers.Dropout(0.2))
model.add(layers.Dense(10))
````

````python
nonsense_pred = model(X_train[:1]).numpy()
tf.nn.softmax(nonsense_pred).numpy()
````

````python
loss_fn = losses.SparseCategoricalCrossentropy(from_logits=True)

loss_fn(y_train[:1], nonsense_pred)
````

````python
optimizer = optimizers.Adam(learning_rate=0.01)
model.compile(optimizer=optimizer, loss=loss_fn, metrics=['accuracy'])
````

````python
model.fit(X_train, y_train, epochs=10)
````

````python
model.evaluate(X_test, y_test)
````

# CIFAR-10 dataset

````python
import tensorflow as tf
from tensorflow.keras import models, layers
from pathlib import Path

DATA_DIR = Path('/home/owner/ml/ObsidianVault') / 'cifar'
CLASS_NAMES = [
    'airplane',
    'automobile',
    'bird',
    'cat',
    'deer',
    'dog',
    'frog',
    'horse',
    'ship',
    'truck',
]
IMAGE_SIZE = (32, 32)
BATCH_SIZE = 32

# The CIFAR images in this workspace are flat PNG files like 0_cat.png.
# Read labels from the filename instead of requiring class subfolders.
image_paths = sorted((DATA_DIR / 'test').glob('*.png'))
if not image_paths:
    raise ValueError(f'No images found in {DATA_DIR / "test"}')

label_to_index = {name: index for index, name in enumerate(CLASS_NAMES)}
labels = [label_to_index[path.stem.split('_')[-1]] for path in image_paths]
paths = [str(path) for path in image_paths]

split_index = max(1, int(len(paths) * 0.8))
train_paths, test_paths = paths[:split_index], paths[split_index:]
train_labels, test_labels = labels[:split_index], labels[split_index:]

def load_image(path, label):
    image = tf.io.read_file(path)
    image = tf.image.decode_png(image, channels=3)
    image = tf.image.resize(image, IMAGE_SIZE)
    image = tf.cast(image, tf.float32) / 255.0
    return image, label

train_ds = tf.data.Dataset.from_tensor_slices((train_paths, train_labels))
train_ds = train_ds.map(load_image, num_parallel_calls=tf.data.AUTOTUNE)
train_ds = train_ds.shuffle(1000).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

test_ds = tf.data.Dataset.from_tensor_slices((test_paths, test_labels))
test_ds = test_ds.map(load_image, num_parallel_calls=tf.data.AUTOTUNE)
test_ds = test_ds.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10),
])

model.compile(
    optimizer='adam',
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    metrics=['accuracy'],
)

print('Starting automated local folder training!')
model.fit(train_ds, epochs=5, validation_data=test_ds)
````
