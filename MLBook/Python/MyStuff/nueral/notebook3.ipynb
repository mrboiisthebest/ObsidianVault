````python
import tensorflow as tf
````

````python
mnist = tf.keras.datasets.mnist

(X_train, y_train), (X_test, y_test) = mnist.load_data()
````

````python
X_train.shape
````

````python
y_train.shape
````

````python
X_train[0]
````

````python
X_train, X_test = X_train / 255.0, X_test / 255 # pre processing making numbers 0 - 1
````

````python
X_train[0]
````

````python
from tensorflow.keras import models, layers, losses, optimizers
````

````python
model = models.Sequential()
model.add(layers.InputLayer((28, 28)))
model.add(layers.Flatten())
model.add(layers.Dense(128, activation='relu'))
model.add(layers.Dropout(0.2))
model.add(layers.Dense(10))
````

````python
nonsense_pred = model(X_train[:1]).numpy()
tf.nn.softmax(nonsense_pred).numpy()
````

````python
loss_fn = losses.SparseCategoricalCrossentropy(from_logits=True)

loss_fn(y_train[:1], nonsense_pred)
````

````python
optimizer = optimizers.Adam(learning_rate=0.01)
model.compile(optimizer=optimizer, loss=loss_fn, metrics=['accuracy'])
````

````python
model.fit(X_train, y_train, epochs=10)
````

````python
model.evaluate(X_test, y_test)
````
